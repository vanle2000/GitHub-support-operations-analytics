# Data Exploration & Cleaning Pipeline
GitHub Support Operations Analytics - Helpdesk Ticket Analysis

This notebook handles initial data inspection and quality checks for the support ticket dataset.

```python
import pandas as pd 

df = pd.read_csv('helpdesk_tickets.csv') 
```

```python
print("Missing Values per Column:") 
print(df.isnull().sum()) 
```

```python
print("\nUnique Labels Found:") 
print(df['Labels'].unique()) 
```

```python
print("\nData Types:") 
print(df.dtypes)
```

```python
# Create a mapping dictionary for standardization 
label_map = { 
    'bug': 'High Priority - Bug', 
    'critical-bug': 'High Priority - Bug', 
    'vulnerability': 'High Priority - Bug', 
    'security': 'High Priority - Bug', 
    'enhancement': 'Feature Request', 
    'feature': 'Feature Request', 
    'question': 'User Inquiry', 
    'help wanted': 'User Inquiry' 
}
```

```python
# Clean Labels
df['Standardized_Label'] = df['Labels'].str.lower().map(label_map).fillna('Other / General')
```

```python
# Set datetime objects for date columns
df['created_at'] = pd.to_datetime(df['created_at']) 
df['closed_at'] = pd.to_datetime(df['closed_at'])
```

```python
# Create Fact Table 
fact_tickets = df[['ticket_id', 'repo_name', 'author_name', 'created_at', 'closed_at', 'Standardized_Label', 'milestone']]
```
